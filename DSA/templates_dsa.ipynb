# Queue Intro

```python
from collections import deque
queue = deque([1,2,3,4,5])

# Add elements:
queue.append(6)
print(queue)

# Peek elements:
print(queue[0])

# Popleft:
queue.popleft()
print(queue)
```

# Sorting

### Insertion sort

```python
def sort_list(unsorted_arr):
    for i in range(len(unsorted_arr)):
        curr = i 
        while (curr > 0)and (unsorted_arr[curr] < unsorted_arr[curr - 1]):
            unsorted_arr[curr],unsorted_arr[curr-1] = unsorted_arr[curr-1],unsorted_arr[curr]
            curr -= 1
    return unsorted_arr

unsorted_arr = [2,1,3,5,4,6,7,9,8]
print(sort_list(unsorted_arr))
```

### Selection sort

```python
def selection_sort(unsorted_arr):
    n = len(unsorted_arr)
    for i in range(n):
        min_index = i 
        for j in range(i, n):
            if unsorted_arr[j] < unsorted_arr[min_index]:
                min_index = j

        # Swap the minimum element with the first element
        unsorted_arr[i], unsorted_arr[min_index] = unsorted_arr[min_index], unsorted_arr[i]
    return unsorted_arr
unsorted_arr = [2,1,3,5,4,6,7,9,8]
print(selection_sort(unsorted_arr))
```

### Bubble sort

```python
def bubble_sort(unsorted_arr):
    n = len(unsorted_arr)
    for i in range(n):
        for j in range(i,n):
            if unsorted_arr[j] < unsorted_arr[i]:
                unsorted_arr[i], unsorted_arr[j] = unsorted_arr[j], unsorted_arr[i]
        
    return unsorted_arr

unsorted_arr = [2,1,3,5,4,6,7,9,8]
print(bubble_sort(unsorted_arr))
```

### Merge sort 

```python
def merge_sort(unsorted_arr):
    n = len(unsorted_arr)

    # Base case: A list of size 1 or 0 is already sorted
    if n <=1 :
      return unsorted_arr
    # Split the list into left and right halves
    midpoint = n // 2
    left_list = merge_sort(unsorted_arr[:midpoint])
    right_list = merge_sort(unsorted_arr[midpoint:])

    result_list = []
    left_pointer, right_pointer = 0, 0

    # Merge the sorted left and right lists with two pointers
    while left_pointer < midpoint or right_pointer < n - midpoint:
        if left_pointer == midpoint:
            # If left list is empty, append element from right
            result_list.append(right_list[right_pointer])
            right_pointer += 1
        elif right_pointer == n - midpoint:
            # If right list is empty, append element from left
            result_list.append(left_list[left_pointer])
            left_pointer += 1
        elif left_list[left_pointer] <= right_list[right_pointer]:
            # Append smaller element from left
            result_list.append(left_list[left_pointer])
            left_pointer += 1
        else:  # Append smaller element from right
            result_list.append(right_list[right_pointer])
            right_pointer += 1

    return result_list
unsorted_arr = [2,1,3,5,4,6,7,9,8]
print(merge_sort(unsorted_arr))
```

### Quick sort

```python
def quick_sort(unsorted_list: list[int], start: int, end: int) -> None:
    # If segment is 1 or 0, it's sorted
    if end - start <= 1:
        return

    # Using last element as the pivot
    pivot = unsorted_list[end - 1]
    start_ptr, end_ptr = start, end - 1

    # Partitioning process
    while start_ptr < end_ptr:
        # Find the next element from the left that is larger than the pivot
        while unsorted_list[start_ptr] < pivot and start_ptr < end_ptr:
            start_ptr += 1

        # Find the next element from the right that is smaller than or equal to the pivot
        while unsorted_list[end_ptr] >= pivot and start_ptr < end_ptr:
            end_ptr -= 1

        # Swap if pointers haven't met
        unsorted_list[start_ptr], unsorted_list[end_ptr] = unsorted_list[end_ptr], unsorted_list[start_ptr]

    # Place pivot in its final position
    unsorted_list[start_ptr], unsorted_list[end - 1] = unsorted_list[end - 1], unsorted_list[start_ptr]

    # Sort left and right of the pivot
    quick_sort(unsorted_list, start, start_ptr)
    quick_sort(unsorted_list, start_ptr + 1, end)
unsorted_arr = [2,1,3,5,4,6,7,9,8]
quick_sort(unsorted_arr,0,len(unsorted_arr)-1)
print(unsorted_arr)
```

# Binary Search

```python
def binary_search(arr, target):
    left = 0
    right = len(arr)-1
    mid = (left + right) //2
    while left < right:
        if arr[mid] == target:
            return True
        elif arr[mid] > target:
            right = mid - 1
        else:
            left = mid+1
    return False
arr = [1,2,3,4,5,6,7,8]
binary_search(arr,4)
```
